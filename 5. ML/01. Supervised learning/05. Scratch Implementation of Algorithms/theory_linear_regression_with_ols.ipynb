# Linear Regression using Ordinary Least Squares (OLS)

## Objective

Instead of repeatedly updating the parameters using Gradient Descent, OLS directly computes the best values of θ using a mathematical formula called the **Normal Equation**.

---

# What is Ordinary Least Squares (OLS)?

Ordinary Least Squares (OLS) is a mathematical method used to find the **best fitting regression line** by minimizing the **sum of squared errors** between actual values and predicted values.

The word means:

- Ordinary → Standard method
- Least → Minimize
- Squares → Square the errors

The objective is

\[
\min \sum (y-\hat y)^2
\]

where

- \(y\) = Actual Value
- \(\hat y\) = Predicted Value

---

# Why is it called Least Squares?

Suppose we have one observation.

Actual Price

```
250
```

Predicted Price

```
230
```

Error

\[
250-230=20
\]

Squared Error

\[
20^2=400
\]

Now suppose all observations have errors.

| Actual | Predicted | Error | Squared Error |
|---------|-----------|-------|---------------|
|200|190|10|100|
|250|230|20|400|
|300|310|-10|100|
|350|340|10|100|

Total Error

\[
100+400+100+100=700
\]

OLS tries to make this value as small as possible.

---

# Problem with Gradient Descent

Gradient Descent finds θ by repeating

```
Predict
↓

Calculate Error
↓

Calculate Gradient
↓

Update θ
↓

Repeat Thousands of Times
```

This takes many iterations.

Question:

Can we directly calculate the best θ?

Answer:

Yes.

Using the **Normal Equation**.

---

# Normal Equation

The Normal Equation is

\[
\boxed{
\theta=(X^TX)^{-1}X^Ty
}
\]

This formula directly gives the best parameters.

No Gradient Descent.

No Learning Rate.

No Iterations.

---

# What does each symbol mean?

## θ (Theta)

Model Parameters

\[
\theta=
\begin{bmatrix}
\theta_0\\
\theta_1\\
\theta_2\\
\vdots
\theta_n
\end{bmatrix}
\]

These are the values we want to find.

---

## X

Feature Matrix

Suppose

| Size | Bedrooms |
|------|-----------|
|1000|2|
|1200|3|
|1500|3|
|1800|4|

After adding the bias column,

\[
X=
\begin{bmatrix}
1&1000&2\\
1&1200&3\\
1&1500&3\\
1&1800&4
\end{bmatrix}
\]

First column of 1's represents the intercept θ₀.

Shape

```
(number_of_samples × number_of_features+1)
```

---

## y

Target Vector

\[
y=
\begin{bmatrix}
200\\
240\\
300\\
360
\end{bmatrix}
\]

Shape

```
(number_of_samples ×1)
```

---

## Xᵀ

Transpose of X

Rows become columns.

Example

If

\[
X=
\begin{bmatrix}
1&2\\
3&4
\end{bmatrix}
\]

then

\[
X^T=
\begin{bmatrix}
1&3\\
2&4
\end{bmatrix}
\]

---

## XᵀX

Multiply

\[
X^TX
\]

Result is a square matrix.

This matrix represents relationships among the input features.

---

## (XᵀX)⁻¹

Inverse Matrix

This is similar to division in matrix algebra.

Example

Normal Numbers

\[
5x=20
\]

Divide both sides

\[
x=\frac{20}{5}
\]

Matrices cannot be divided directly.

Instead,

we multiply by the inverse.

\[
AX=B
\]

becomes

\[
X=A^{-1}B
\]

That is why

\[
(X^TX)^{-1}
\]

appears in the formula.

---

## Xᵀy

Multiply the transpose of X with the target vector.

This combines the input features with their actual outputs.

---

# Professional Example

Suppose we want to predict employee salary.

Dataset

| Experience | Salary |
|------------|---------|
|1|45|
|2|55|
|3|65|
|4|75|

---

## Step 1

Create X

\[
X=
\begin{bmatrix}
1&1\\
1&2\\
1&3\\
1&4
\end{bmatrix}
\]

First column = Bias

Second column = Experience

---

## Step 2

Create y

\[
y=
\begin{bmatrix}
45\\
55\\
65\\
75
\end{bmatrix}
\]

---

## Step 3

Compute

\[
X^T=
\begin{bmatrix}
1&1&1&1\\
1&2&3&4
\end{bmatrix}
\]

---

## Step 4

Multiply

\[
X^TX=
\begin{bmatrix}
4&10\\
10&30
\end{bmatrix}
\]

---

## Step 5

Inverse

\[
(X^TX)^{-1}
=
\frac{1}{20}
\begin{bmatrix}
30&-10\\
-10&4
\end{bmatrix}
=
\begin{bmatrix}
1.5&-0.5\\
-0.5&0.2
\end{bmatrix}
\]

---

## Step 6

Multiply

\[
X^Ty=
\begin{bmatrix}
240\\
650
\end{bmatrix}
\]

---

## Step 7

Final Parameters

\[
\theta
=
(X^TX)^{-1}X^Ty
\]

\[
=
\begin{bmatrix}
35\\
10
\end{bmatrix}
\]

Therefore

\[
\boxed{\hat y=35+10x}
\]

Interpretation

- Intercept = 35
- Every extra year of experience increases salary by approximately 10 units.

---

# NumPy Implementation

```python
import numpy as np

X = np.array([[1],
              [2],
              [3],
              [4]])

y = np.array([45,55,65,75])

# Add bias column
X_b = np.c_[np.ones((len(X),1)), X]

theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y

print(theta)
```

Output

```
[35. 10.]
```

---

# Prediction

```python
new_employee = np.array([1,5])   # [Bias, Experience]

salary = new_employee @ theta

print(salary)
```

Output

```
85
```

Meaning

Employee with 5 years of experience is predicted to earn approximately **85 units**.

# Gradient Descent vs OLS

| Feature | Gradient Descent | OLS |
|----------|------------------|-----|
|Iterations|Yes|No|
|Learning Rate|Required|Not Required|
|Normal Equation|No|Yes|
|Speed on Small Data|Slower|Very Fast|
|Speed on Large Data|Better|Slower (matrix inversion is expensive)|
|Feature Scaling|Recommended|Usually not required for correctness|
|Best For|Large datasets|Small to medium datasets|

---

# Advantages of OLS

- Direct mathematical solution.
- No iterations.
- No learning rate tuning.
- Finds the exact optimum when the normal equation is applicable.
- Simple implementation using linear algebra.

---

# Limitations

- Computing \((X^TX)^{-1}\) becomes expensive for datasets with many features.
- If \(X^TX\) is singular (non-invertible), the normal equation cannot be applied directly.
- Requires more memory for very large datasets.

---

# Summary

```
Training Data
      ↓
Create Feature Matrix X
      ↓
Create Target Vector y
      ↓
Compute Xᵀ
      ↓
Compute XᵀX
      ↓
Compute (XᵀX)⁻¹
      ↓
Compute Xᵀy
      ↓
θ = (XᵀX)⁻¹Xᵀy
      ↓
Best Parameters Found
      ↓
Prediction
```
