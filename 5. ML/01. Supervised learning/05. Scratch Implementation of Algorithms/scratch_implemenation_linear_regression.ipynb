```python
# Linear Regression Implementation from scratch
import numpy as np

class LinearRegression:
    def __init__(self, learning_rate = 0.01, n_iteration = 1000):
        self.bias = None
        self.weights = None
        self.lr_rate = learning_rate
        self.n_iter = n_iteration

    def fit(self, X, y): # X_train, y_train
        m, n = X.shape # m = number of samples, n = number of features
        
        # Step 1 -- initialize
        self.bias = 0
        self.weights = np.zeros(n) # will create array of zeros = [0,0,0,....]

        # Gradient Descent
        for i in range(self.n_iter):
            # Step 2 -- calculate y_pred
            y_pred = self.bias + np.dot(X, self.weights)
    
            # Step 3 -- calculate Gradient 
            db = (1/m) * np.sum(y_pred - y)
            dw = (1/m) * np.dot(X.T, (y_pred - y))
    
            # Step 4 -- Convergence Theorem -- Update parameters
            self.bias -= self.lr_rate * db
            self.weights -= self.lr_rate * dw
    
    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)
print(model.bias)
print(model.weights)
```
