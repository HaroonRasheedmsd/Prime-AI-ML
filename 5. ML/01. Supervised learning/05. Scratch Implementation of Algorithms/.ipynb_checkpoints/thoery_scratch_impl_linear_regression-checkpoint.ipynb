# Linear Regression From Scratch using Gradient Descent

## Objective

Implement Linear Regression from scratch without using `sklearn.LinearRegression`.

The model will learn the relationship between input features (X) and target values (y) by minimizing the prediction error using **Gradient Descent**.

---

# Step 1: Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

- NumPy → Mathematical calculations
- Matplotlib → Visualization

---

# Step 2: Prepare Dataset

Suppose we have the following dataset.

| House Size (1000 sqft) | Price (100k $) |
|-------------------------|----------------|
|1|2|
|2|4|
|3|6|
|4|8|
|5|10|

```python
X = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])
```

---

# Step 3: Understand the Equation

Linear Regression tries to learn the equation

\[
\hat y = \theta_0 + \theta_1X
\]

where

- θ₀ = Intercept (Bias)
- θ₁ = Weight (Slope)
- ŷ = Predicted value

Initially, the computer doesn't know θ₀ and θ₁.

---

# Step 4: Random Initialization

Start with random values.

```python
theta0 = 0
theta1 = 0
```

At this point,

\[
\hat y = 0
\]

The model knows nothing yet.

---

# Step 5: Prediction Function

Create a function that predicts output.

```python
def predict(X, theta0, theta1):
    return theta0 + theta1 * X
```

Example

```python
y_pred = predict(X, theta0, theta1)
```

---

# Step 6: Calculate Error

Error tells how far prediction is from actual value.

\[
Error = \hat y - y
\]

```python
error = y_pred - y
```

---

# Step 7: Cost Function

We use Mean Squared Error (MSE).

\[
J(\theta)=\frac1{2m}\sum(\hat y-y)^2
\]

where

- m = number of samples

```python
def cost_function(X, y, theta0, theta1):

    m = len(y)

    y_pred = predict(X, theta0, theta1)

    cost = (1/(2*m))*np.sum((y_pred-y)**2)

    return cost
```

Purpose:

Measure how bad the model is.

Lower cost = Better model.

---

# Step 8: Calculate Gradients

Gradient tells how parameters should change.

Gradient for θ₀

\[
\frac{\partial J}{\partial \theta_0}
=
\frac1m\sum(\hat y-y)
\]

Gradient for θ₁

\[
\frac{\partial J}{\partial \theta_1}
=
\frac1m\sum(\hat y-y)x
\]

Implementation

```python
def gradients(X,y,theta0,theta1):

    m=len(y)

    y_pred=predict(X,theta0,theta1)

    error=y_pred-y

    d_theta0=(1/m)*np.sum(error)

    d_theta1=(1/m)*np.sum(error*X)

    return d_theta0,d_theta1
```

---

# Step 9: Update Parameters

Gradient Descent Formula

\[
\theta=\theta-\alpha\times gradient
\]

where

- α = Learning Rate

```python
learning_rate = 0.01
```

Update

```python
theta0 = theta0 - learning_rate*d_theta0

theta1 = theta1 - learning_rate*d_theta1
```

After updating,

the line becomes slightly better.

---

# Step 10: Repeat Until Convergence

Now repeat

1. Predict
2. Calculate Error
3. Calculate Cost
4. Calculate Gradients
5. Update Parameters

This loop continues until cost stops decreasing.

---

# Step 11: Complete Training Function

```python
def gradient_descent(X,y,theta0,theta1,learning_rate,epochs):

    cost_history=[]

    for i in range(epochs):

        d_theta0,d_theta1=gradients(X,y,theta0,theta1)

        theta0=theta0-learning_rate*d_theta0

        theta1=theta1-learning_rate*d_theta1

        cost=cost_function(X,y,theta0,theta1)

        cost_history.append(cost)

    return theta0,theta1,cost_history
```

---

# Step 12: Train Model

```python
theta0=0
theta1=0

learning_rate=0.01

epochs=1000

theta0,theta1,cost_history=gradient_descent(
    X,
    y,
    theta0,
    theta1,
    learning_rate,
    epochs
)
```

Now the model has learned

- Best θ₀
- Best θ₁

---

# Step 13: Final Equation

```python
print(theta0)
print(theta1)
```

Suppose

```
theta0 = 0.02

theta1 = 1.99
```

Equation becomes

\[
\hat y =0.02+1.99X
\]

---

# Step 14: Predict New Data

Suppose

House Size = 6

```python
new_x=6

prediction=predict(new_x,theta0,theta1)

print(prediction)
```

Output

```
11.96
```

Approximately

```
12
```

---

# Step 15: Visualize Regression Line

```python
plt.scatter(X,y,label="Actual Data")

plt.plot(X,predict(X,theta0,theta1),
         color="red",
         label="Regression Line")

plt.legend()

plt.show()
```

Blue Points → Actual Data

Red Line → Learned Line

---

# Step 16: Plot Cost Reduction

```python
plt.plot(cost_history)

plt.xlabel("Iterations")

plt.ylabel("Cost")

plt.title("Gradient Descent")

plt.show()
```

The graph should continuously decrease, showing that the model is learning.

---

# Complete Algorithm

```
Import Libraries
        ↓
Load Dataset
        ↓
Initialize θ₀ and θ₁
        ↓
Predict Output
        ↓
Calculate Error
        ↓
Calculate Cost
        ↓
Compute Gradients
        ↓
Update Parameters
        ↓
Repeat Until Convergence
        ↓
Final Model Ready
        ↓
Predict New Data
```

---

# Time Complexity

- Prediction → O(n)
- Cost Function → O(n)
- Gradient Calculation → O(n)
- Total Training → O(n × epochs)

---

# Advantages

- Simple and easy to understand.
- Fast for small datasets.
- Interpretable model.
- Good baseline algorithm.

---

# Limitations

- Assumes linear relationship.
- Sensitive to outliers.
- Requires feature scaling for faster convergence.
- Cannot capture complex nonlinear patterns.

---

# Interview Questions

### Why initialize parameters randomly?

Because the model has no prior knowledge and needs a starting point.

### Why use Mean Squared Error?

Because it penalizes large errors more heavily and provides a smooth, differentiable cost function.

### Why Gradient Descent?

To minimize the cost function by finding the optimal values of θ₀ and θ₁.

### Why subtract the gradient?

Because the gradient points in the direction of increasing cost; subtracting it moves toward lower cost.

### What happens after training?

The learned values of θ₀ and θ₁ are used to predict outputs for new unseen data.

---
