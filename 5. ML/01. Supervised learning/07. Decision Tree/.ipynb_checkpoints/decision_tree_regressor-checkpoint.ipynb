```python
# Dicision Tree Regressor To Solve a perticular Problem

# For this we take Diabetes Dataset from sklearn and
# Predict Diabetes Progression

import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
```

```python
df = load_diabetes(as_frame=True).frame
```

```python
df.head()
```

```python
X = df.drop("target", axis =1)
y = df["target"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, 
    y,
    test_size=0.3,
    random_state=42
)
```

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(max_depth=7, min_samples_leaf=20)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import r2_score, mean_squared_error

y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)

print("MSE train: ", mean_squared_error(y_train, y_pred_train))
print("MSE test: ", mean_squared_error(y_test, y_pred_test))

print("r^2 train: ", r2_score(y_train, y_pred_train))
print("r^2 test: ", r2_score(y_test, y_pred_test))
```

```python
# As in the above it was extream Overfitting like :
# MSE train: 0.0 erro  &  r^2 train: 1.0 error 


# To Plot it
from sklearn.tree import plot_tree

plt.figure(figsize=(18, 10))

plot_tree(
    model,
    feature_names=X.columns,
    filled=True
)

plt.tight_layout()

# as model was overfitted and tree size also show that tree is very big now: 
# Two option we can apply:
    # 1. Prepruning
    # 2. Postpruning 

# lets apply prepruning as simpler also so will apply above training model portion
# as we change the hyperparameter values i.e:(max_depth=7, min_samples_leaf=20
# and r^2 values little come closer: shown in above avaluation portion
```
