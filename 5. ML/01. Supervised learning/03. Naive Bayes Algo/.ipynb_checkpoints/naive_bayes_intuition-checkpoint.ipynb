# Naive Bayes

## Definition

**Naive Bayes** is a **Supervised Machine Learning Classification Algorithm** that predicts the most likely class of a new data point using **Bayes' Theorem** and **Probability**.

---

# What Problem Does Naive Bayes Solve?

Naive Bayes is used for **Classification Problems** where the output is a category instead of a numerical value.

### Examples

- Spam / Not Spam
- Disease A / Disease B
- Positive / Negative Review
- Fraud / Not Fraud
- Loan Approved / Rejected
- Fake News / Real News

---

# Intuition

Instead of drawing the best line like **Linear Regression**, Naive Bayes predicts by calculating **probabilities**.

It asks one simple question:

> **Which class has the highest probability?**

The class with the highest probability becomes the prediction.

---

# Real-Life Example 1: Medical Diagnosis

Suppose a patient has:

- Fever
- Cough
- Body Pain

A doctor thinks:

- Flu?
- COVID?
- Viral Fever?

The doctor chooses the disease with the **highest probability**.

Naive Bayes works in the same way.

---

# Real-Life Example 2: Spam Detection

A new email contains:

- Win
- Lottery
- Prize
- Money

Naive Bayes calculates:

- Probability(Spam | Email)
- Probability(Not Spam | Email)

If the spam probability is higher,

**Prediction = Spam**

---

# What Does Naive Bayes Learn?

During training, Naive Bayes learns probabilities from the training dataset.

Example:

Suppose there are 100 emails.

- Spam = 70
- Not Spam = 30

It learns:

- P(Spam) = 70 / 100
- P(Not Spam) = 30 / 100

It also learns probabilities of features.

Example:

- P(Money | Spam)
- P(Offer | Spam)
- P(Lottery | Spam)

These probabilities are stored for future predictions.

---

# Training Phase

During training:

1. Read the training data.
2. Calculate prior probabilities.
3. Calculate conditional probabilities for every feature.
4. Store these probabilities.

Training ends here.

---

# Prediction Phase

When a new data point arrives:

1. Calculate the probability for each class.
2. Compare all probabilities.
3. Select the class with the highest probability.

This becomes the final prediction.

---

# Bayes' Theorem

Naive Bayes is based on **Bayes' Theorem**.

\[
P(Class|Data)=\frac{P(Data|Class)\times P(Class)}{P(Data)}
\]

Where

- **P(Class | Data)** → Posterior Probability
- **P(Data | Class)** → Likelihood
- **P(Class)** → Prior Probability
- **P(Data)** → Evidence

During prediction, the denominator is the same for every class, so Naive Bayes compares only:

\[
P(Data|Class)\times P(Class)
\]

---

# Why is it Called "Naive"?

Naive Bayes assumes that **all input features are independent of each other.**

Example:

A patient has

- Fever
- Cough

In reality,

Fever and cough may be related.

Naive Bayes assumes they are **independent**.

This assumption is called the **Naive Assumption**.

Although this assumption is not always true, Naive Bayes often performs very well.

---

# Complete Working Flow

```
Training Data
      │
      ▼
Calculate Probabilities
      │
      ▼
Store Probabilities
      │
      ▼
New Data
      │
      ▼
Apply Bayes' Theorem
      │
      ▼
Calculate Probability of Every Class
      │
      ▼
Choose Highest Probability
      │
      ▼
Prediction
```

---

# One-Line Intuition

> **Naive Bayes predicts the class with the highest probability.**

---

# Easy Definition

**Naive Bayes is a supervised machine learning classification algorithm that learns probabilities from training data and uses Bayes' Theorem to predict the class with the highest probability while assuming that all input features are independent of each other.**

---

# Types of Naive Bayes

All Naive Bayes algorithms use the same **Bayes' Theorem**.

The only difference is **the type of data they work with.**

```
Naive Bayes
      │
      ├── Gaussian Naive Bayes
      ├── Multinomial Naive Bayes
      ├── Bernoulli Naive Bayes
      ├── Categorical Naive Bayes
      └── Complement Naive Bayes
```

---

# How to Choose the Correct Type?

Choose the Naive Bayes type based on the **feature values**.

| Feature Type | Naive Bayes Type |
|--------------|------------------|
| Continuous Numbers | Gaussian NB |
| Counts / Frequencies | Multinomial NB |
| Binary (0/1, Yes/No) | Bernoulli NB |
| Categories | Categorical NB |
| Imbalanced Text Data | Complement NB |

---

# 1. Gaussian Naive Bayes

## Definition

Gaussian Naive Bayes is used when the features are **continuous numerical values**.

### Examples

- Height
- Weight
- Temperature
- Salary
- Age
- Blood Pressure

---

## Assumption

It assumes that the feature values follow a **Normal (Gaussian) Distribution**.

---

## Real-Life Example

Predict disease using:

- Temperature
- Blood Pressure
- Heart Rate

Since these are continuous values,

**Use Gaussian Naive Bayes.**

---

## Applications

- Medical Diagnosis
- Weather Prediction
- Risk Prediction
- Sensor Data
- Biological Measurements

---

## When to Use

- Continuous numerical features.

---

## When Not to Use

- Words
- Categories
- Binary values

---

# 2. Multinomial Naive Bayes

## Definition

Multinomial Naive Bayes is used when the features represent **counts or frequencies**.

### Examples

- Word appears 5 times.
- Word appears 10 times.
- Product purchased 4 times.

---

## Real-Life Example

Email

| Word | Count |
|------|------|
| Money | 5 |
| Win | 3 |
| Offer | 2 |

These are word frequencies.

**Use Multinomial Naive Bayes.**

---

## Applications

- Spam Detection
- Text Classification
- News Classification
- Sentiment Analysis
- Document Classification

---

## When to Use

- Count Data
- Frequency Data
- Bag of Words
- TF-IDF Features

---

## When Not to Use

- Continuous numerical values

---

# 3. Bernoulli Naive Bayes

## Definition

Bernoulli Naive Bayes is used when every feature is **Binary**.

Possible values:

- 0 / 1
- Yes / No
- True / False

---

## Real-Life Example

Instead of counting words, only check whether a word exists.

| Word | Present |
|------|---------|
| Money | 1 |
| Lottery | 0 |
| Prize | 1 |
| Offer | 0 |

Since the values are binary,

**Use Bernoulli Naive Bayes.**

---

## Applications

- Spam Detection
- Binary Features
- Survey Data
- Medical Symptoms (Present / Absent)

---

## When to Use

- Binary Features
- Yes / No Data
- True / False Data

---

## When Not to Use

- Continuous Values
- Count Data

---

# 4. Categorical Naive Bayes

## Definition

Categorical Naive Bayes is used when the features are **categorical values**.

### Examples

- Red
- Blue
- Green

- Toyota
- Honda
- BMW

- Male
- Female

---

## Real-Life Example

Predict whether a customer will buy a car.

Features:

- Color
- Brand
- Transmission

These are categories.

**Use Categorical Naive Bayes.**

---

## Applications

- Customer Prediction
- Marketing
- Product Recommendation
- Survey Analysis

---

## When to Use

- Categorical Features

---

## When Not to Use

- Continuous Numerical Values
- Count Data

---

# 5. Complement Naive Bayes

## Definition

Complement Naive Bayes is a modified version of **Multinomial Naive Bayes** designed for **imbalanced datasets**, especially text classification.

---

## Real-Life Example

Email Dataset

- Spam = 100
- Not Spam = 9000

The classes are highly imbalanced.

Complement Naive Bayes handles this better than Multinomial Naive Bayes.

---

## Applications

- Large Text Classification
- Spam Detection
- News Classification
- Imbalanced Text Data

---

## When to Use

- Imbalanced Text Datasets

---

## When Not to Use

- Continuous Numerical Data
- Binary Features

---

# Comparison Table

| Type | Feature Type | Example | Common Applications |
|------|--------------|----------|---------------------|
| Gaussian NB | Continuous Numbers | Height, Weight, Temperature | Medical Diagnosis |
| Multinomial NB | Counts / Frequencies | Word Counts | Spam Detection |
| Bernoulli NB | Binary (0/1) | Word Present / Not Present | Email Filtering |
| Categorical NB | Categories | Color, Brand, City | Customer Prediction |
| Complement NB | Imbalanced Count Data | Large Text Data | Text Classification |

---

# Advantages

- Simple and easy to understand.
- Very fast training.
- Very fast prediction.
- Works well with small datasets.
- Handles many input features efficiently.
- Excellent for text classification.
- Requires less training data than many algorithms.

---

# Disadvantages

- Assumes features are independent.
- Performance decreases when features are highly correlated.
- May perform worse than more complex models on difficult datasets.
- Can produce zero probabilities for unseen feature values (handled later using Laplace Smoothing).

---

# Industry Applications

- Spam Detection
- Medical Diagnosis
- Fraud Detection
- Product Categorization
- News Classification
- Sentiment Analysis
- Customer Support Ticket Classification

---

# Summary

Naive Bayes is a fast and simple supervised classification algorithm that predicts the class with the highest probability using Bayes' Theorem. The choice of Naive Bayes type depends on the type of input features:

- Continuous Numbers → Gaussian NB
- Counts/Frequencies → Multinomial NB
- Binary Values → Bernoulli NB
- Categories → Categorical NB
- Imbalanced Text Data → Complement NB
