# Working of Naive Bayes Algorithm

## Goal

The goal of Naive Bayes is to **predict the class with the highest probability** for a new data point.

---

# Complete Working Steps

```
Training Data
      │
      ▼
Calculate Prior Probability
      │
      ▼
Calculate Likelihood
      │
      ▼
Store Probabilities
      │
      ▼
New Data
      │
      ▼
Apply Bayes' Theorem
      │
      ▼
Calculate Posterior Probability
      │
      ▼
Compare All Classes
      │
      ▼
Choose Highest Probability
      │
      ▼
Final Prediction
```

---

# Example Dataset

Suppose we have the following dataset.

| Weather | Wind | Play |
|----------|------|------|
| Hot | Weak | Yes |
| Hot | Weak | Yes |
| Cold | Weak | No |
| Hot | Strong | No |
| Cold | Strong | Yes |

Features

- Weather
- Wind

Target

- Play (Yes / No)

We want to predict

```
Weather = Hot
Wind = Strong
```

Will the person play?

---

# Step 1 : Calculate Prior Probability

## What is Prior Probability?

Prior Probability is the probability of each class before looking at the input features.

Formula

\[
P(Class)=\frac{\text{Number of rows of that class}}{\text{Total rows}}
\]

---

### Calculate P(Yes)

Number of Yes = 3

Total rows = 5

\[
P(Yes)=\frac{3}{5}=0.6
\]

---

### Calculate P(No)

Number of No = 2

Total rows = 5

\[
P(No)=\frac{2}{5}=0.4
\]

---

## Interpretation

Before looking at Weather and Wind,

there is already a 60% chance of Yes.

---

# Step 2 : Calculate Likelihood

## What is Likelihood?

Likelihood means

> **How likely is a feature value given a class?**

Formula

\[
P(Feature|Class)
\]

---

### Weather Probabilities

#### For Class = Yes

Rows having Yes

| Weather |
|----------|
| Hot |
| Hot |
| Cold |

Hot appears = 2

Cold appears = 1

Therefore

\[
P(Hot|Yes)=\frac{2}{3}
\]

\[
P(Cold|Yes)=\frac{1}{3}
\]

---

#### For Class = No

Rows having No

| Weather |
|----------|
| Cold |
| Hot |

Hot appears = 1

Cold appears = 1

Therefore

\[
P(Hot|No)=\frac{1}{2}
\]

\[
P(Cold|No)=\frac{1}{2}
\]

---

### Wind Probabilities

#### For Class = Yes

Rows

Weak

Weak

Strong

Weak = 2

Strong = 1

Therefore

\[
P(Weak|Yes)=\frac{2}{3}
\]

\[
P(Strong|Yes)=\frac{1}{3}
\]

---

#### For Class = No

Rows

Weak

Strong

Weak = 1

Strong = 1

Therefore

\[
P(Weak|No)=\frac{1}{2}
\]

\[
P(Strong|No)=\frac{1}{2}
\]

---

# Step 3 : Apply Bayes' Theorem

Suppose the new data is

```
Weather = Hot

Wind = Strong
```

Now calculate probability for every class.

---

## Formula

\[
P(Class|Features)\propto P(Class)\times P(Feature_1|Class)\times P(Feature_2|Class)
\]

The denominator

\[
P(Features)
\]

is the same for every class.

Therefore we ignore it while comparing classes.

---

# Step 4 : Calculate Probability for Class = Yes

Known values

\[
P(Yes)=\frac35
\]

\[
P(Hot|Yes)=\frac23
\]

\[
P(Strong|Yes)=\frac13
\]

Multiply

\[
P(Yes|Hot,Strong)
\propto
\frac35
\times
\frac23
\times
\frac13
\]

\[
=
\frac{2}{15}
=
0.133
\]

---

# Step 5 : Calculate Probability for Class = No

Known values

\[
P(No)=\frac25
\]

\[
P(Hot|No)=\frac12
\]

\[
P(Strong|No)=\frac12
\]

Multiply

\[
P(No|Hot,Strong)
\propto
\frac25
\times
\frac12
\times
\frac12
\]

\[
=
\frac{1}{10}
=
0.10
\]

---

# Step 6 : Compare Both Probabilities

| Class | Probability |
|--------|------------|
| Yes | 0.133 |
| No | 0.10 |

Since

```
0.133 > 0.10
```

Prediction is

```
Play = Yes
```

---

# Why Do We Multiply?

Because of the **Naive Assumption**.

Naive Bayes assumes

```
Weather

and

Wind
```

are independent.

Therefore

\[
P(Weather,Wind|Class)
=
P(Weather|Class)
\times
P(Wind|Class)
\]

Without this assumption,

the calculation would become much more complicated.

---

# What Does Naive Bayes Learn During Training?

Naive Bayes does **not** learn a line or equation.

Instead, it stores probabilities.

Example

```
P(Yes)

P(No)

P(Hot|Yes)

P(Cold|Yes)

P(Hot|No)

P(Cold|No)

P(Weak|Yes)

P(Strong|Yes)

P(Weak|No)

P(Strong|No)
```

These probabilities are stored in memory.

---

# What Happens During Prediction?

When new data arrives,

Naive Bayes

1. Reads stored probabilities.
2. Applies Bayes' Theorem.
3. Calculates probability for every class.
4. Compares all probabilities.
5. Chooses the class with the highest probability.

---

# Understanding the Four Terms of Bayes' Theorem

Bayes' Theorem

\[
P(Class|Data)
=
\frac
{P(Data|Class)\times P(Class)}
{P(Data)}
\]

---

## 1. Posterior Probability

\[
P(Class|Data)
\]

Meaning

Probability of a class **after seeing the data**.

Example

```
P(Yes | Hot, Strong)
```

This is what we want to calculate.

---

## 2. Prior Probability

\[
P(Class)
\]

Meaning

Probability of a class before looking at any feature.

Example

```
P(Yes)=3/5

P(No)=2/5
```

---

## 3. Likelihood

\[
P(Data|Class)
\]

Meaning

Probability of observing the feature values if the class is known.

Example

```
P(Hot|Yes)

P(Strong|Yes)
```

---

## 4. Evidence

\[
P(Data)
\]

Meaning

Overall probability of observing the given data.

Example

```
P(Hot, Strong)
```

During prediction,

Evidence is the same for every class,

so Naive Bayes ignores it while comparing classes.

---

# Why Can We Ignore Evidence?

Suppose

```
Yes Probability =

0.20 / 0.40

No Probability =

0.10 / 0.40
```

The denominator

```
0.40
```

is the same.

Therefore,

only the numerators need to be compared.

---

# Important Note

Naive Bayes compares **relative probabilities**, not necessarily the final normalized probabilities.

The class with the **largest value** is selected.

---

# Summary

1. Read the training data.
2. Calculate Prior Probability.
3. Calculate Likelihood for every feature.
4. Store all probabilities.
5. Receive new input.
6. Apply Bayes' Theorem.
7. Ignore Evidence because it is constant.
8. Compute Posterior Probability for every class.
9. Compare all probabilities.
10. Choose the class with the highest probability.

---

# One-Line Definition of Important Terms

| Term | Definition |
|------|------------|
| Prior Probability | Probability of a class before seeing the data. |
| Likelihood | Probability of the feature values given the class. |
| Posterior Probability | Probability of the class after seeing the data. |
| Evidence | Overall probability of observing the given data. |
| Naive Assumption | Assumes all input features are independent of each other. |

---

# Key Takeaways

- Naive Bayes is a probability-based classification algorithm.
- It learns probabilities from the training data.
- It uses Bayes' Theorem to make predictions.
- It assumes features are independent.
- It predicts the class with the highest posterior probability.
