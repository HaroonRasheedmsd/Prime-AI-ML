# Laplace Smoothing (Add-One Smoothing)

## Why Do We Need Laplace Smoothing?

Naive Bayes calculates probabilities from the training data.

Sometimes a feature value **never appears** with a particular class.

In that case,

its probability becomes

\[
P(Feature|Class)=0
\]

Since Naive Bayes multiplies all probabilities,

even one zero makes the entire probability zero.

This can lead to an incorrect prediction.

Laplace Smoothing solves this problem.

---

# The Zero Probability Problem

Suppose our training dataset is

| Weather | Wind | Play |
|----------|------|------|
| Hot | Weak | Yes |
| Hot | Weak | Yes |
| Cold | Weak | No |
| Hot | Strong | No |
| Cold | Strong | Yes |

Now imagine a new feature

```
Weather = Rainy
```

But

Rainy never appeared during training.

Then

\[
P(Rainy|Yes)=0
\]

and

\[
P(Rainy|No)=0
\]

Now calculate

\[
P(Yes|Rainy)
\]

Naive Bayes multiplies

```
Prior × Likelihood
```

Since

```
Likelihood = 0
```

Final probability becomes

```
0
```

This is called the **Zero Probability Problem**.

---

# Why Is This a Problem?

Suppose

```
P(Yes)=0.60

P(Rainy|Yes)=0
```

Then

\[
0.60 \times 0 = 0
\]

Even though

Yes may actually be the correct class,

Naive Bayes completely rejects it.

This is not a good behavior.

---

# Solution

Instead of allowing probability to become zero,

we add **1** to every feature count.

This technique is called

## Laplace Smoothing

or

## Add-One Smoothing

---

# Simple Idea

Instead of saying

```
Rainy occurred 0 times
```

we pretend

```
Rainy occurred 1 time
```

This prevents zero probabilities.

---

# Formula Without Smoothing

\[
P(Feature|Class)
=
\frac{\text{Feature Count}}
{\text{Total Class Count}}
\]

---

# Formula With Laplace Smoothing

\[
P(Feature|Class)
=
\frac{\text{Feature Count}+1}
{\text{Total Class Count}+\text{Number of Possible Feature Values}}
\]

Notice two changes

- Add **1** to the numerator.
- Add the **number of possible feature values** to the denominator.

---

# Example

Suppose

Weather has three possible values

```
Hot

Cold

Rainy
```

For class = Yes

Training data

| Weather |
|----------|
| Hot |
| Hot |
| Cold |

Total rows

```
3
```

Rainy appears

```
0
```

---

# Without Laplace Smoothing

\[
P(Rainy|Yes)
=
\frac03
=
0
\]

This becomes a problem.

---

# With Laplace Smoothing

Feature Count

```
0
```

Add one

```
0+1=1
```

Total rows

```
3
```

Possible Weather values

```
3
```

Therefore

\[
P(Rainy|Yes)
=
\frac{0+1}{3+3}
=
\frac16
=
0.167
\]

Notice

The probability is **small**,

but it is **not zero**.

Now Naive Bayes can still make a prediction.

---

# Another Example

Suppose

Vocabulary contains

```
Money

Offer

Prize

Lottery
```

Total words

```
4
```

During training

Word

```
Lottery
```

never appeared in Spam emails.

Without smoothing

\[
P(Lottery|Spam)=0
\]

With Laplace Smoothing

\[
P(Lottery|Spam)
=
\frac{0+1}
{\text{Total Spam Words}+4}
\]

Now the probability becomes small,

instead of zero.

---

# Why Add One?

Adding one means

> "Give every possible feature at least one chance."

Instead of saying

```
Impossible
```

Naive Bayes says

```
Very unlikely,
but still possible.
```

---

# When Do We Use Laplace Smoothing?

Laplace Smoothing is mainly used when

- A feature value never appeared during training.
- Zero probabilities occur.
- Working with text classification.
- Using Multinomial Naive Bayes.
- Using Bernoulli Naive Bayes.

---

# Advantages

- Removes zero probabilities.
- Improves prediction for unseen feature values.
- Makes the model more robust.
- Prevents multiplication from becoming zero.

---

# Disadvantages

- Slightly changes the original probabilities.
- If too much smoothing is used, probabilities become less accurate.

---

# Does Scikit-Learn Use Laplace Smoothing?

Yes.

In Scikit-Learn,

Naive Bayes models have a parameter

```
alpha
```

```
alpha = 1
```

means

Laplace Smoothing is applied.

Example

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB(alpha=1.0)
```

If

```
alpha = 0
```

No smoothing is applied.

---

# Laplace Smoothing vs No Smoothing

| Without Laplace Smoothing | With Laplace Smoothing |
|---------------------------|------------------------|
| Probability can become 0 | Probability never becomes 0 |
| Model may fail on unseen values | Handles unseen values better |
| Less robust | More robust |
| Multiplication may become zero | Multiplication remains valid |

---

# Summary

1. Naive Bayes calculates probabilities from the training data.
2. Sometimes a feature never appears with a class.
3. Its probability becomes zero.
4. One zero makes the entire prediction zero.
5. Laplace Smoothing solves this by adding **1** to every feature count.
6. This produces a small probability instead of zero.
7. Scikit-Learn uses the **alpha** parameter to control smoothing.

| Term | Definition |
|------|------------|

| Alpha | A smoothing parameter used in Scikit-Learn Naive Bayes models. |

# Key Takeaways

- Laplace Smoothing prevents zero probabilities.
- It makes Naive Bayes more reliable.
