# Before Learning Cross Validation

Before learning **Cross Validation**, we must first understand another dataset called **Validation Data**.

Most beginners think Machine Learning only has

- Training Data
- Testing Data

But in real projects, there is another important dataset called

> **Validation Data**

Without understanding Validation Data, Cross Validation will not make sense.

---

# What is Validation Data?

Validation Data is a part of the dataset that is used to **check whether our model is learning correctly while we are building it.**

It is **not used to train the model.**

It is **not used for the final evaluation.**

Its only job is

> **Help us choose the best model before the final testing.**

---

# What is Cross Validation?

Cross Validation is a technique where

the dataset is divided into multiple parts,

and every part gets a chance to become the Validation Data.

Instead of checking the model once,

we check it many times.

Finally,

we take the average performance.

---

# Simple Definition

> **Cross Validation is a model evaluation technique in which different parts of the dataset become the Validation Data one by one, and the final performance is calculated by averaging all the validation results.**

---

# Why do we use Cross Validation?

Because

one Validation split

may be

```
Easy

or

Hard
```

Cross Validation removes this problem.

Instead of trusting

one split,

it trusts

all splits.

So the result becomes much more reliable.

---

# Summary

- **Training Data** teaches the model.
- **Validation Data** helps us improve and select the best model.
- **Testing Data** is used only once for the final evaluation.
- We cannot evaluate using Training Data because the model has already seen it.
- We should not repeatedly use Testing Data because it causes **Data Leakage**.
- Validation Data solves this problem, but a single validation split can still give misleading results.
- **Cross Validation** solves this by using multiple validation splits and averaging the results, giving a much more reliable estimate of model performance.

# What is K-Fold Cross Validation?

K-Fold Cross Validation is a technique where

the dataset is divided into **K equal parts (called folds).**

Then,

each fold becomes the Validation Data **one by one.**

Finally,

all validation scores are averaged.

That average is considered the model's actual performance.

---

# Simple Definition

> **K-Fold Cross Validation is a technique where every part of the dataset gets one chance to become the Validation Data, and the average of all validation results becomes the final performance.**

# Advantages of K-Fold Cross Validation

- Uses every sample for both training and validation.
- Gives more reliable performance.
- Reduces the effect of random splitting.
- Better model selection.
- Better hyperparameter tuning.
- Widely used in industry.

---

# Disadvantages

- Slower than a single train-validation split.
- The model must be trained multiple times.
- Can be expensive for very large datasets.

---

# Summary

- **K-Fold Cross Validation** divides the dataset into **K equal folds**.
- Each fold becomes the **Validation Data exactly one time**.
- The remaining folds are used for training.
- The process repeats until every fold has been used as the validation set.
- The validation scores from all rounds are averaged to get the final model performance.
- This approach reduces the effect of random data splitting and provides a much more reliable estimate than using only one validation set.
- In practice, **5-Fold** and **10-Fold Cross Validation** are the most commonly used choices because they balance reliability and computation time.

# Types (Ways) of Cross Validation

So far, we have learned **K-Fold Cross Validation**.

But K-Fold is not the only type.

There are different types of Cross Validation, and each one is used for a different situation.

# 1. Hold-Out Validation

## What is Hold-Out Validation?

Hold-Out Validation is the simplest validation technique.

We split the dataset only **once**.

Example

```
1000 Samples

↓

80%

Training

↓

20%

Validation
```

The model is trained only one time.

Then we check the performance only one time.

```
100 Students
```

```
80 Students

Training
```

```
20 Students

Validation
```

Train the model

↓

```
91%
```

Done.

No more training.

---

## Advantages

- Very fast.
- Easy to understand.
- Good for very large datasets.

---

## Disadvantages

Performance depends on

```
One Random Split
```

Sometimes

```
91%
```

Sometimes

```
84%
```

Sometimes

```
95%
```

Not very reliable.

---

# 2. K-Fold Cross Validation

We already studied this.

Example

```
100 Samples

↓

5 Folds
```

Every fold becomes Validation once.

Then

Average Accuracy

is calculated.

---

## Best For

- Small datasets
- Medium datasets
- Most ML projects

# 3. Stratified K-Fold Cross Validation

## Why was it introduced?

Let's understand with a real example.

---

Suppose we are building a

```
Fraud Detection Model
```

Dataset

```
1000 Transactions
```

Among them

```
950 Normal

50 Fraud
```

That means

```
95%

Normal
```

```
5%

Fraud
```

---

Now suppose we use normal K-Fold.

One fold may accidentally become

```
200 Samples

↓

198 Normal

↓

2 Fraud
```

Another fold may become

```
200 Samples

↓

190 Normal

↓

10 Fraud
```

Now every fold has different class ratios.

This affects model performance.

---

## Solution

Stratified K-Fold keeps the same class ratio in every fold.

Original Dataset

```
95%

Normal
```

```
5%

Fraud
```

Every Fold

also becomes

```
95%

Normal
```

```
5%

Fraud
```

Now every fold represents the original dataset.

This makes evaluation much more reliable.

---

## Where is it used?

Mainly

```
Classification Problems
```

Especially

```
Imbalanced Datasets
```

Example

- Fraud Detection
- Disease Prediction
- Spam Detection

---

# 4. Leave-One-Out Cross Validation (LOOCV)

This is an extreme version of K-Fold.

Suppose

```
100 Samples
```

```
5 Folds
```

```
100 Folds
```

```
Only 1 Sample
```

---

### Round 1

Training

```
99 Samples
```

Validation

```
1 Sample
```

---

### Round 2

Training

```
99 Samples
```

Validation

```
Another Sample
```

Repeat

```
100 Times
```

---

## Advantages

Almost every sample is used for training.

Very accurate estimate.

---

## Disadvantages

Very slow.

Imagine

```
1,000,000 Samples
```

You would train

```
1,000,000 Models
```

Not practical.

---

# 5. Repeated K-Fold

Normal K-Fold

```
5 Folds

↓

5 Models
```

Repeated K-Fold

```
5 Folds

↓

Repeat Again

↓

Repeat Again

↓

Repeat Again
```

Example

```
5-Fold

Repeated

3 Times
```

Now

```
15 Models
```

This reduces randomness even more.

# Comparison

| Type | Best For | Speed |
|------|----------|-------|
| Hold-Out | Large datasets | Very Fast |
| K-Fold | General ML | Medium |

# Pipeline in Scikit-Learn

## Why was Pipeline introduced?

Let's understand with an example.

Suppose

we are building

a kNN model.

Our steps are

```
Load Data

↓

Scale Data

↓

Train Model

↓

Predict
```

Everything is fine.

---

Now

we decide to use

```
5-Fold Cross Validation
```

Every fold has

different

Training Data

and

Validation Data.

Question

Should Feature Scaling happen

before

or

after

the split?

Correct answer

```
After Split
```

Because

the scaler should learn only from the Training Data.

---

# The Problem Without Pipeline

Imagine

Fold 1

```
Scale Whole Dataset ❌

↓

Split Data

↓

Train
```

Now

the Validation Data

has already influenced

the scaler.

This is

```
Data Leakage
```

The model receives information

that it should never have seen.

Your Cross Validation score becomes overly optimistic.

---

# Correct Process

For every fold

```
Training Data

↓

Fit Scaler

↓

Transform Training Data

↓

Transform Validation Data

↓

Train Model

↓

Validate
```

This entire process

must repeat

for every fold.

Doing this manually

is difficult.

---

# Solution

Scikit-Learn introduced

```
Pipeline
```

---

# What is a Pipeline?

A Pipeline is a tool that

connects multiple Machine Learning steps

into one workflow.

Instead of writing

```
Scaling

↓

Training

↓

Prediction
```

separately,

Pipeline performs them

automatically

in the correct order.

---

# Without Pipeline

```
Split Data

↓

Scale Training

↓

Scale Testing

↓

Train

↓

Predict
```

You write every step manually.

---

# With Pipeline

```
Split Data

↓

Pipeline

↓

Scaling

↓

Training

↓

Prediction
```

One object

handles everything.

---

# Why do we use Pipeline?

- Cleaner code.
- Fewer mistakes.
- Prevents Data Leakage.
- Works perfectly with Cross Validation.
- Works perfectly with Hyperparameter Tuning (`GridSearchCV`, `RandomizedSearchCV`).
- Easier to maintain in production.

---

# Industry Workflow

Almost every real Machine Learning project follows a workflow similar to this:

```
Collect Data

↓

Preprocess Data

↓

Train-Test Split

↓

Create Pipeline

↓

Feature Scaling

↓

Model

↓

Cross Validation

↓

Hyperparameter Tuning

↓

Select Best Model

↓

Train Final Model

↓

Evaluate on Test Data

↓

Deploy Model
```

Notice that **Testing Data is used only once at the end**, after all tuning and model selection are complete.

---

# Summary

- **Hold-Out Validation** uses one train-validation split and is fast but less reliable.
- **K-Fold Cross Validation** uses every fold as validation once and averages the results.
- **Stratified K-Fold** keeps the class distribution the same in every fold, making it ideal for classification with imbalanced data.
- **LOOCV** uses one sample for validation in each round and is very accurate but computationally expensive.
- **Repeated K-Fold** repeats K-Fold multiple times to reduce randomness further.
- **Hyperparameter Tuning** uses Cross Validation to compare different hyperparameter values (such as different values of **K** in kNN) and select the one with the best average validation performance.
- A **Pipeline** automatically performs preprocessing and model training in the correct order, prevents **Data Leakage**, and integrates seamlessly with Cross Validation and Hyperparameter Tuning.
