# Ways / Categories / Types of kNN

---

# Introduction

kNN is one algorithm, but it can work in different ways depending on the problem.

These different ways are called **categories**, **types**, or **variations** of kNN.

---

# 1. kNN Classification

## What is it?

kNN Classification is used when the output is a **category (label)**.

Instead of predicting a number, it predicts a class.

### Examples

| Input | Output |
|--------|--------|
| Email | Spam / Not Spam |
| Patient Data | Healthy / Sick |
| Student Data | Pass / Fail |
| Image | Cat / Dog |

---

## How does it work?

It finds the **K nearest neighbors** and counts which class appears the most.

The class with the highest vote becomes the prediction.

### Example

Suppose

K = 5

Nearest neighbors are

```
Pass

Pass

Fail

Pass

Fail
```

Votes

```
Pass = 3

Fail = 2
```

Prediction

```
Pass
```

This process is called **Majority Voting**.

---

# 2. kNN Regression

## What is it?

kNN Regression is used when the output is a **number** instead of a category.

Instead of majority voting,

it calculates the **average** of the nearest neighbors.

---

### Example

Suppose we want to predict house price.

Nearest neighbors have prices

```
220K

230K

210K
```

Prediction

```
(220 + 230 + 210)

-------------

3

= 220K
```

So,

Predicted Price = **220K**

---

# Difference Between Classification and Regression

| kNN Classification | kNN Regression |
|---------------------|----------------|
| Predicts Category | Predicts Number |
| Uses Majority Voting | Uses Average |
| Example: Spam Detection | Example: House Price |

---

# 3. Distance Metrics

## What is a Distance Metric?

A distance metric is a formula that tells us

> **"How close or how far two data points are."**

Smaller distance means

```
More Similar
```

Larger distance means

```
Less Similar
```

Different datasets may require different distance metrics.

---

# (A) Euclidean Distance

## What is it?

Euclidean Distance is the **straight-line distance** between two points.

It is the **default** and most commonly used distance metric in kNN.

### Real-Life Example

Imagine two houses on a map.

The shortest straight road between them is the Euclidean Distance.

```
A ●

 \

  \

   \

    ● B
```

---

### Formula

For two features

```
Distance = √[(x₂ − x₁)² + (y₂ − y₁)²]
```

---

### When to use?

- Continuous numerical data
- Most ML datasets
- Default choice for kNN

---

# (B) Manhattan Distance

## What is it?

Manhattan Distance measures distance by moving only

- Left
- Right
- Up
- Down

Just like moving on city roads.

---

### Real-Life Example

Suppose you are driving in a city.

You cannot drive through buildings.

You must follow roads.

```
□ □ □ □

□ A □ □

□ □ □ □

□ □ B □
```

You travel block by block.

---

### Formula

```
Distance = |x₂ − x₁| + |y₂ − y₁|
```

---

### When to use?

- Grid-based movement
- City maps
- Some high-dimensional datasets

---

# (C) Minkowski Distance

## What is it?

Minkowski Distance is a **general distance formula**.

It can behave like different distance metrics by changing one parameter.

---

### Special Cases

If

```
p = 1
```

It becomes

```
Manhattan Distance
```

If

```
p = 2
```

It becomes

```
Euclidean Distance
```

---

### Why use it?

Because one formula can represent multiple distance metrics.

---

# (D) Hamming Distance

## What is it?

Hamming Distance counts

> **How many values are different.**

It is mainly used for

- Binary Data
- Categorical Data
- Text
- DNA Sequences

---

### Example

```
101101

111001
```

Compare each position

```
1 = 1 ✔

0 ≠ 1 ✖

1 = 1 ✔

1 ≠ 0 ✖

0 = 0 ✔

1 = 1 ✔
```

Different positions = **2**

So

```
Hamming Distance = 2
```

---

# (E) Cosine Similarity (Basic)

## What is it?

Cosine Similarity measures

> **How similar the direction of two vectors is.**

It does **not** focus on distance.

It focuses on similarity.

---

### Example

Two students study different hours.

Student A

```
2

4

6
```

Student B

```
4

8

12
```

The values are different,

but the pattern is almost the same.

Cosine Similarity says

```
Very Similar
```

---

### Where is it used?

- Recommendation Systems
- NLP
- Search Engines
- Text Similarity

---

# Which Distance Metric Should You Use?

| Distance Metric | Best For |
|-----------------|----------|
| Euclidean | Numerical Data |
| Manhattan | Grid or City Movement |
| Minkowski | General Purpose |
| Hamming | Binary & Categorical Data |
| Cosine Similarity | Text & Recommendation Systems |

# 4. Choosing the Value of K

The value of **K** decides

> **How many neighbors the algorithm should consider before making a prediction.**

Choosing the right K is very important.

---

# Small K

Example

```
K = 1

K = 3
```

### Advantages

- Captures local patterns.
- Can detect small groups.

### Disadvantages

- Sensitive to noise.
- May overfit.

---

# Large K

Example

```
K = 15

K = 25
```

### Advantages

- More stable predictions.
- Less affected by noise.

### Disadvantages

- May ignore important local patterns.
- Can underfit.

---

# Odd vs Even K

Usually we choose an **odd value** of K.

Example

```
K = 3

K = 5

K = 7
```

Reason

Odd K reduces the chance of a tie in classification.

---

### Example

If

```
K = 4

Pass

Pass

Fail

Fail
```

Votes

```
Pass = 2

Fail = 2
```

Tie occurs.

---

If

```
K = 5

Pass

Pass

Pass

Fail

Fail
```

Votes

```
Pass = 3

Fail = 2
```

No tie.

---

# How to Choose the Best K?

There is no fixed value.

Usually we

- Try different K values.
- Compare model accuracy.
- Select the K that gives the best performance on validation/testing data.

---

# 5. Weighting Methods

Sometimes every neighbor is treated equally.

Sometimes closer neighbors are given more importance.

This creates two weighting methods.

---

## Uniform Weight

Every neighbor has the same importance.

Example

```
Neighbor 1 = 1 Vote

Neighbor 2 = 1 Vote

Neighbor 3 = 1 Vote
```

---

## Distance Weight

Closer neighbors get higher importance.

Farther neighbors get lower importance.

Example

```
Closest Neighbor

★★★★★

Far Neighbor

★
```

This often improves prediction quality.

---

# 6. Feature Scaling

## Why is Feature Scaling Important?

kNN compares distances.

If one feature has much larger values than another,

it dominates the distance calculation.

---

### Example

| Age | Salary |
|------|---------|
| 25 | 50000 |
| 30 | 70000 |

Salary values are much larger.

Without scaling,

Salary affects the distance much more than Age.

---

### Solution

Scale the features before training.

Common methods

- Standardization
- Normalization

---

# 7. Curse of Dimensionality

## What is it?

When the number of features becomes very large,

almost every data point appears far from every other point.

As a result,

finding true nearest neighbors becomes difficult.

---

### Example

Imagine

- 2 Features → Easy to compare.
- 100 Features → Very difficult to find similar points.

Prediction quality decreases.

---

# Summary

- kNN works in two main categories:
  - **Classification** (predicts categories using majority voting)
  - **Regression** (predicts numbers using the average of neighbors)
- Distance metrics decide how similarity is measured.
- Euclidean Distance is the default choice for most numerical datasets.
- Choosing the correct **K** is essential for good performance.
- **Odd values of K** are generally preferred to reduce ties.
- **Feature Scaling** is important because kNN is distance-based.
- With too many features, kNN suffers from the **Curse of Dimensionality**.
