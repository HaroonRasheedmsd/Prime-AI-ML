# Dimensionality Reduction

## 1. Where Does Dimensionality Reduction Fit?

Dimensionality Reduction is one of the major areas of **Unsupervised Learning**.

```text
                    UNSUPERVISED LEARNING
                           │
             ┌─────────────┼─────────────┐
             │             │             │
        Clustering    Dimensionality   Association
                       Reduction         Rules
             │             │
          Find groups   Reduce dimensions
                        while preserving
                       useful information
```

You are currently studying:

> **Dimensionality Reduction**

The important idea is:

> **Dimensionality reduction is not simply deleting columns. It is creating a smaller representation of the same data while preserving useful information or structure.**

---

# 2. What Problem Are We Actually Solving?

Imagine an e-commerce dataset containing:

| Customer | Age | Income | Time on Site | Pages Viewed | Products Viewed | Cart Items | Purchases |
| -------- | --- | ------ | ------------ | ------------ | --------------- | ---------- | --------- |
| A        | 22  | 40k    | 15           | 8            | 12              | 2          | 1         |
| B        | 45  | 90k    | 35           | 20           | 35              | 5          | 4         |

Suppose we have:

```text
100 features
```

Every customer is represented in a:

```text
100-dimensional space
```

Humans cannot visualize 100 dimensions.

Even worse, some features may be:

* Redundant
* Correlated
* Noisy
* Irrelevant
* Expensive to process

For example:

```text
Income
Annual Salary
Monthly Salary
Yearly Earnings
```

These may contain almost the same information.

So we ask:

> **Can I represent these 100 features using, say, 10 dimensions while retaining most of the useful structure?**

That's the fundamental idea of dimensionality reduction.

* 2D

```text
50 Features
      ↓
Dimensionality Reduction
      ↓
2 Features
      ↓
Plot
```

---

Suppose:

```text
10,000 features
```

Training a model can become expensive.

If we can represent the data using:

```text
100 dimensions
```

computation can become much easier.

---

## 4.3 Remove Redundancy

Suppose:

```text
Height
Weight
BMI
Waist
```

contain overlapping information.

Dimensionality reduction can compress correlated information.

---

## 4.4 Reduce Noise

Some dimensions may contain mostly noise.

Reducing dimensions can sometimes remove some of that noise.

---

## 4.5 Help Downstream ML

You can do:

```text
Original Data
     ↓
Dimensionality Reduction
     ↓
Clustering / Classification / Regression
```

For example:

```text
100 Features
     ↓
PCA
     ↓
10 Components
     ↓
K-Means
     ↓
Customer Segments
```

# 5. Feature Selection vs Feature Extraction

This distinction is extremely important.

## Feature Selection

Feature selection means choosing existing features.

For example:

```text
Original Features:

Age
Income
Height
Weight
Salary
```

After feature selection:

```text
Age
Income
Salary
```

You **keep the original columns**.

---

## Feature Extraction

Feature extraction creates new features from the old ones.

For example:

```text
Original Features:

Age
Income
Height
Weight
...
```

becomes:

```text
Component 1
Component 2
Component 3
```

These components are **new mathematical combinations** of the original features.

Many dimensionality-reduction techniques such as **PCA** perform feature extraction.

### Remember

```text
Feature Selection
       ↓
Keep existing features

Feature Extraction
       ↓
Create new features
```

# 8. PCA — Principal Component Analysis

PCA is one of the most important dimensionality-reduction techniques.

## Intuition



Imagine these points:

```text
       •
      •
     •
    •
   •
  •
```

The points have a strong direction.

Instead of describing every point using:

```text
X
Y
```

we can find a new direction that captures most of the variation.

That direction becomes:

> **Principal Component 1 (PC1)**

Then we find another direction perpendicular to PC1:

> **Principal Component 2 (PC2)**

If PC1 captures almost everything important:

```text
X, Y
 ↓
PC1
```

We reduced:

```text
2D → 1D
```

---

# 9. What Is PCA Really Doing?

PCA asks:

> **"What direction through my data captures the maximum variance?"**

Then:

> **"What is the next best direction, perpendicular to the first?"**

Then:

> **"Continue until I have enough components."**

So:

```text
Original Features
       ↓
Find Direction of Maximum Variance
       ↓
PC1
       ↓
Find Next Best Perpendicular Direction
       ↓
PC2
       ↓
PC3
       ↓
...
```

The core idea:

> **PCA finds new directions that capture as much variation in the data as possible.**

---

# 10. Why Does PCA Use Variance?

Suppose we have:

```text
Feature A

10
11
12
13
14
```

and:

```text
Feature B

10
50
11
100
5
```

Feature B varies much more.

Variance measures how spread out the data is.

PCA assumes:

> **Directions with high variance contain important structure or information.**

Therefore, PCA tries to preserve as much variance as possible.

---

# 11. PCA Mathematics

Suppose our dataset is:

```text
X
```

with:

```text
n samples × p features
```

For example:

```text
X = 1000 × 50
```

Suppose we want:

```text
1000 × 5
```

Therefore:

```text
50 dimensions → 5 dimensions
```

---

# 12. Step 1 — Standardization

Before PCA, we usually standardize numerical features.

The formula is:

$$
z = \frac{x-\mu}{\sigma}
$$

where:

* $x$ = original value
* $\mu$ = mean
* $\sigma$ = standard deviation
* $z$ = standardized value

For example:

```text
Age       → 20–60
Income    → 20,000–200,000
```

Income has a much larger numerical scale.

Without scaling, features with larger numerical values can have a disproportionate influence on PCA.

Therefore:

```text
Original Data
      ↓
StandardScaler
      ↓
Scaled Data
      ↓
PCA
```

---

# 13. Step 2 — Covariance Matrix

PCA examines how features vary together.

For two features:

$$
Cov(X,Y)
$$

If they increase together:

```text
Height ↑
Weight ↑
```

they tend to have positive covariance.

If one increases while the other decreases:

```text
Price ↑
Demand ↓
```

they may have negative covariance.

PCA uses the covariance matrix:

$$
\Sigma
$$

to understand relationships between features.

For example:

$$
\Sigma =
\begin{bmatrix}
Cov(X_1,X_1) & Cov(X_1,X_2) & Cov(X_1,X_3) \\
Cov(X_2,X_1) & Cov(X_2,X_2) & Cov(X_2,X_3) \\
Cov(X_3,X_1) & Cov(X_3,X_2) & Cov(X_3,X_3)
\end{bmatrix}
$$

The diagonal contains the variance of each feature.

The off-diagonal values describe covariance between different features.

---

# 14. Step 3 — Eigenvectors and Eigenvalues

This is the mathematical heart of PCA.

PCA solves:

$$
\Sigma v = \lambda v
$$

where:

* $v$ = eigenvector → **direction**
* $\lambda$ = eigenvalue → **amount of variance captured by that direction**

Think:

```text
Eigenvector = Direction
Eigenvalue  = Importance
```

Suppose PCA produces:

```text
PC1 → eigenvalue = 8.0
PC2 → eigenvalue = 1.5
PC3 → eigenvalue = 0.3
PC4 → eigenvalue = 0.2
```

Total variance:

$$
8 + 1.5 + 0.3 + 0.2 = 10
$$

Therefore:

$$
\frac{8}{10}=80\%
$$

PC1 explains **80%**.

PC2:

$$
\frac{1.5}{10}=15\%
$$

PC3:

$$
\frac{0.3}{10}=3\%
$$

PC4:

$$
\frac{0.2}{10}=2\%
$$

Therefore:

```text
PC1 = 80%
PC2 = 15%
PC3 = 3%
PC4 = 2%
```

---

# 15. Explained Variance

If we keep PC1 and PC2:

$$
80\% + 15\% = 95\%
$$

Therefore:

```text
Original:

4 dimensions
```

becomes:

```text
Reduced:

2 dimensions
```

while retaining approximately:

```text
95% of the variance
```

This is the heart of PCA.

---

# 16. Step 4 — Project the Data

Once PCA finds the new directions, we project the original data onto those directions.

Conceptually:

$$
Z = XW
$$

where:

* $X$ = original data
* $W$ = selected principal-component directions
* $Z$ = transformed/reduced data

Example:

```text
Original:

100 features
     ↓
PCA
     ↓
10 components
```

Therefore:

```text
1000 × 100
     ↓
1000 × 10
```

The number of samples remains the same.

Only the number of dimensions is reduced.

---

# 17. How Do I Choose the Number of Components?

This is one of the most important practical questions.

You don't normally say:

> "I'll always use 2 components."

Instead, use **explained variance**.

For example:

```text
Component       Explained Variance

PC1             45%
PC2             25%
PC3             12%
PC4              7%
PC5              4%
PC6              3%
PC7              2%
PC8              1%
PC9              1%
PC10             0.5%
```

Cumulative variance:

```text
PC1       → 45%
PC1 + PC2 → 70%
PC1–PC3   → 82%
PC1–PC4   → 89%
PC1–PC5   → 93%
PC1–PC6   → 96%
```

If your target is around **95%**, choose:

```text
6 components
```

instead of 10.

# 23. When Should We NOT Use PCA?

PCA is not magic.

## 23.1 Mostly Categorical Data

For example:

```text
Gender
Country
Payment Method
Product Category
```

Ordinary PCA is designed around numerical relationships.

Don't blindly apply PCA to categorical variables.

---

## 23.2 Interpretability Is Critical

Original features:

```text
Age
Income
Credit Score
Debt
```

After PCA:

```text
PC1
PC2
PC3
```

PC1 is harder to explain to a business stakeholder.

For example:

```text
PC1 =
0.42 × Age
+ 0.51 × Income
- 0.18 × Debt
+ 0.37 × Credit Score
+ ...
```

Therefore, if interpretability is extremely important, feature selection may be better.

---

## 23.3 Nonlinear Structure

PCA is fundamentally a linear method.

Suppose the data has a complex nonlinear structure:

```text
       XXXXX
     XX
    X
   X
    X
     XX
       XXXXX
```

A straight-line technique like PCA may not capture this structure effectively.

This is where nonlinear methods such as:

```text
t-SNE
UMAP
```

can be useful.

# 24. t-SNE

**t-SNE = t-distributed Stochastic Neighbor Embedding**

Forget the mathematical name initially.

Its intuition is:

> **"Keep points that were close together in high-dimensional space close together in the low-dimensional representation."**

Suppose we have:

```text
Customer Data
      ↓
100 dimensions
      ↓
t-SNE
      ↓
2 dimensions
```

We might see:

```text
   ● ● ●          ○ ○ ○
  ● ● ●            ○ ○
   ● ●              ○ ○


             ▲ ▲ ▲
            ▲ ▲ ▲
```

You can visually identify local groups.

---

# 25. What Is t-SNE Mainly Used For?

t-SNE is mainly used for:

* Visualization
* Exploratory data analysis
* Understanding local structure
* Exploring embeddings
* Finding visually separated groups

Typical workflow:

```text
High-dimensional embeddings
          ↓
         t-SNE
          ↓
          2D
          ↓
         Plot
          ↓
Understand local structure
```

---

# 26. Important t-SNE Warning

t-SNE is primarily a **visualization technique**, not simply a replacement for PCA.

Be careful when interpreting the plot.

The following are not necessarily meaningful:

* Exact distance between distant clusters
* Cluster size
* Global geometry
* Density differences

A visually separated t-SNE cluster does not automatically mean that the original data contains a true business or scientific cluster.

Use it for:

> **Exploration, not blind conclusions.**

# 29. UMAP

**UMAP = Uniform Manifold Approximation and Projection**

UMAP is another nonlinear dimensionality-reduction technique.

Its basic intuition is:

> **"Preserve important neighborhood relationships while creating a lower-dimensional representation."**

Typical workflow:

```text
High-dimensional data
        ↓
       UMAP
        ↓
      2D / 3D
        ↓
Visualization
```

UMAP is commonly used with:

* NLP embeddings
* Image embeddings
* Biological data
* Customer representations
* High-dimensional feature vectors

---

# 30. UMAP vs t-SNE

Both are nonlinear techniques.

```text
High-dimensional data
        │
        ├───────────────┐
        ↓               ↓
      t-SNE            UMAP
        │               │
        ↓               ↓
      2D map          2D map
```

In practical applications:

* t-SNE is especially focused on local neighborhood visualization.
* UMAP can provide a useful balance between local and broader structure.
* UMAP is often more scalable for larger datasets.

# 33. LDA — Linear Discriminant Analysis

Here LDA means:

> **Linear Discriminant Analysis**

Important:

LDA is different from PCA because LDA uses **labels**.

Therefore:

```text
PCA
↓
Unsupervised
↓
Does not need y
```

while:

```text
LDA
↓
Supervised
↓
Uses X + y
```

---

# 34. PCA vs LDA Intuition

Suppose we have two classes:

```text
Class A = Healthy
Class B = Disease
```

PCA asks:

> **Which direction captures the most overall variance?**

LDA asks:

> **Which direction separates the known classes best?**

Therefore:

```text
PCA → Maximum Variance

LDA → Maximum Class Separation
```

---

# 35. LDA Mathematical Intuition

LDA tries to find a projection where:

```text
Between-class variation
        ↓
       HIGH
```

while:

```text
Within-class variation
        ↓
       LOW
```

In simple terms:

> **Points belonging to the same class should be close together, while different classes should be far apart.**

Conceptually:

```text
Original Data

   ● ● ●
  ● ● ●

                ▲ ▲ ▲
               ▲ ▲ ▲
```

LDA searches for a direction where the classes become well separated after projection.

---

# 36. LDA Has a Special Constraint

If you have:

```text
C classes
```

the maximum number of LDA components is:

$$
C - 1
$$

For example:

```text
2 classes → maximum 1 component
3 classes → maximum 2 components
4 classes → maximum 3 components
```

So LDA is strongly connected to the number of classes.

---

# 37. PCA vs LDA

| PCA                                   | LDA                                  |
| ------------------------------------- | ------------------------------------ |
| Unsupervised                          | Supervised                           |
| Does not require labels               | Requires labels                      |
| Maximizes variance                    | Maximizes class separation           |
| General dimensionality reduction      | Class-aware dimensionality reduction |
| Components based on feature variation | Components based on class structure  |

Remember:

> **PCA doesn't know what the classes are.**

> **LDA knows the classes and tries to separate them.**

Don't confuse these two.

## Dimensionality Reduction

> **"How can I represent this data using fewer dimensions?"**

```text
100D
 ↓
10D
```

---

Answers:

> **"Which observations appear to belong to the same group?"**

They can be combined:

```text
100 Features
     ↓
PCA
     ↓
10 Components
     ↓
K-Means
     ↓
Customer Groups
```

# 45. The Mathematical Picture

For **PCA**, remember this chain:

```text
Data X
 ↓
Standardize
 ↓
Covariance Matrix
 ↓
Eigenvectors + Eigenvalues
 ↓
Rank Components by Eigenvalue
 ↓
Choose k Components
 ↓
Project Data
 ↓
Reduced X
```

Mathematically:

$$
X
\rightarrow
\Sigma
\rightarrow
(\lambda,v)
\rightarrow
W_k
\rightarrow
Z = XW_k
$$

Where:

```text
X  = Original data
Σ  = Covariance matrix
λ  = Eigenvalues
v  = Eigenvectors
W  = Selected principal directions
Z  = Reduced representation
```

# 46. Common Mistakes

## Mistake 1 — Forgetting to Scale Before PCA

Bad:

```python
pca.fit_transform(X)
```

when features have very different scales.

Better:

```python
X_scaled = scaler.fit_transform(X)
X_pca = pca.fit_transform(X_scaled)
```

---

## Mistake 2 — Data Leakage

Don't do:

```python
scaler.fit_transform(X)
```

before splitting into train and test data.

Instead:

```text
Split
 ↓
Fit preprocessing on training data
 ↓
Transform training data
 ↓
Transform test data
```

---

## Mistake 3 — Assuming More Components Are Always Better

The goal is reduction.

You need to find a useful trade-off between:

```text
Information Retained
        vs
Number of Dimensions
```

---

## Mistake 4 — Treating t-SNE as Ground Truth

A t-SNE plot is an exploratory visualization.

Different settings can produce different results.

Don't automatically conclude:

```text
Separated visually = true business clusters
```

---

## Mistake 5 — Expecting PCA Components to Be Easily Interpretable

A component is usually a combination of many original features.

For example:

```text
PC1 =
0.42 × Age
+ 0.51 × Income
- 0.18 × Debt
+ 0.37 × Credit Score
+ ...
```

Therefore PC1 may not have a simple human-readable meaning.

# 47. Final Decision Framework

When you receive a real-world dataset, don't immediately choose PCA.

Ask these questions:

```text
1. How many features do I have?
        ↓
2. Are the features numerical?
        ↓
3. Are they correlated/redundant?
        ↓
4. Do I need compression?
        ↓
5. Do I need visualization?
        ↓
6. Do I have labels?
        ↓
7. Is the structure linear or nonlinear?
        ↓
8. What information do I want to preserve?
```

Then choose:

```text
Correlated numerical data
        ↓
       PCA
```

```text
Visualization + local structure
        ↓
      t-SNE
```

```text
Visualization + nonlinear structure + larger datasets
        ↓
      UMAP
```

```text
Known labels + class separation
        ↓
       LDA
```

# 49. Apply the 6 Questions to Dimensionality Reduction

## WHAT?

Reduce high-dimensional data into fewer dimensions.

## WHY?

To:

* Simplify data
* Visualize data
* Compress information
* Remove redundancy
* Sometimes reduce noise
* Improve computational efficiency
* Create useful representations for downstream ML

## HOW?

Transform the original feature space into a lower-dimensional representation.

## WHICH?

```text
PCA
→ Linear / general-purpose numerical reduction

t-SNE
→ Local-structure visualization

UMAP
→ Nonlinear visualization / reduction

LDA
→ Supervised class separation
```

## WHEN?

When:

* The number of features is large
* Features are redundant
* Data is difficult to visualize
* Computation is expensive
* You need a compact representation

## WHERE?

Common applications include:

* E-commerce
* Finance
* Computer Vision
* NLP
* Biology
* Manufacturing
* Recommendation Systems
* Sensor Analytics

# 50. Final Mental Model

The most important thing to remember:

> **Dimensionality Reduction = taking a complicated high-dimensional dataset and finding a smaller representation that preserves the information or structure we actually care about.**

Remember these three sentences:

> **PCA asks: "What directions explain the most variation?"**

> **t-SNE / UMAP ask: "How can I represent neighborhood structure in a low-dimensional map?"**

> **LDA asks: "What projection best separates my known classes?"**
