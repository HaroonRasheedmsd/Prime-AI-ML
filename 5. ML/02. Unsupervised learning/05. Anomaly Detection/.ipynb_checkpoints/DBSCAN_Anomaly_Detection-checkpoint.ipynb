# Anomaly Detection using DBSCAN 

```python
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
```

```python
# generate data
X, y = make_moons(
    n_samples=500,
    noise=0.1,
    random_state=42
)
```

```python
# Scale data
scale = StandardScaler()
X_scaled = scale.fit_transform(X)
```

```python
# Visualize
sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:,1])
```

```python
# DBSCAN 

dbscan = DBSCAN(
    eps=0.5,
    min_samples=5
)

labels = dbscan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:,1], c=labels)

# as here only one cluster form no outliers Due to 2 Problems:
# 1. we gave eps_val bigger | 2. we gave min_sample tow low 
```

```python
# DBSCAN : little decrease eps_val or increase min_sample but we will:
# decrease eps_val not min_sample as there are only two features

dbscan = DBSCAN(
    eps=0.18,
    min_samples=5
)

labels = dbscan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:,1], c=labels)
```
