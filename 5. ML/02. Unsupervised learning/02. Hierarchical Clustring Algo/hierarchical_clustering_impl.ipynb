```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()

X = iris.data
y = iris.target
```

```python
# Scale the values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Visualize 
# sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 2], c=y) # generally we don't write the expectation in Real data which is c=y
sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 2])
```

```python
# Before we apply Agglomerative Algo we have to figure out that:
# How many Clusters we need for tha we should use Dendrogram

# For Dendrogram we have to use a SciPy Python Library:
# that help us in Fundamental Algos for scientific computing in Python


from scipy.cluster.hierarchy import linkage, dendrogram

# first we have to create Linkage matrix: that store data for Dendrogram
Z = linkage(X_scaled, method="ward")

# plot the dendrom gram
plt.figure(figsize=(12, 6))
dendrogram(Z)
plt.xlabel("samples")
plt.ylabel("distance")
plt.title("Dendrogram fro hierarchical clustering")
```

```python
# Agglomerative Clustering

from sklearn.cluster import AgglomerativeClustering

agg = AgglomerativeClustering(n_clusters=2)
labels = agg.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 2], c=labels)
```
