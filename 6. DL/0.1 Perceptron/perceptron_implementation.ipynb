# Implementation of Perceptron

```python
import torch
import torch.nn as nn
```

```python
# create inputs 
torch.manual_seed(42) # as this is used like to create same output as run same code anywhere again and again like in ML(random_state)
inputs = torch.tensor([1.0, 2.0, 3.0])
```

```python
neuron = nn.Linear(in_features=3, out_features=1)
```

```python
output = neuron(inputs)
```

```python
output
```

```python
# Verify ... Formula
neuron.weight @ inputs + neuron.bias
```

```python
# can see weightss 
neuron.weight
```

```python
# can seee bias
neuron.bias
```
