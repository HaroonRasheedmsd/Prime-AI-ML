```python
# Lets do Web Scraping: using module like: i.request, ii.BeatifulSoup, iii.selinium, there are others also

import requests
URL = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)
res
```

```python
# some methods of requests module
print(res.status_code) # HTTP status code if 200 means ok
print(res.text) # html content of the page
print(res.content) # Raw bytes of the response
print(res.headers) # Metadata like content type, server info etc

# also like this
if res.status_code == 200:
    print(res.headers)
```

```python
# To store data in files 
with open("scraped_data/data.html", "w", encoding="utf-8") as f:
    f.write(res.text)
```

```python
# BeautifulSoup
#!pip install beautifulsoup4 # to install beautifulsoup
#!pip install lxml # to install lxml for fast parsing

# it is python library used to parse html and xml documents. it helps us to extract data from webpages after dwonloading
# them with requests. so it create parse tree from data from which we can extract text, attributes, links, table, etc
from bs4 import BeautifulSoup
with open("scraped_data/data.html", "r") as f:
    html_content = f.read()

# now create object of Class(BeautifulSoup)
soup = BeautifulSoup(html_content, "lxml")
soup
```

```python
# BeautifulSoup Methods etc
# methods: 
# find() gives the first occurance of the elements of anythings
soup.find("h3") 
# Direclty access first occurance
soup.h1

# find_all() : gives all the occurance of the elemnets 
all_h3 = soup.find_all("h3")

# this is like list so we can apply loop also like
for h3 in all_h3:
    print(h3)

# if we want to extract country name only means subdata for that get_text(), find_parent(), 
for h3 in all_h3:
    print(h3.get_text(strip = True)) # here get_text() get data and strip = True removes spaces 
    print(h3.find_parent("div")["class"]) # find_parent() get parent of child and also using [] can give you attributes

# also find_next(): used to find next in documnet and find_previous(): used to find previouse of the document

# CSS selectors is used to access css attribute etc we use i.select() to select many like list and ii.select_one for one only
# To access "Attributes" we use get() for single and for all we use .attrs
for h3 in all_h3:
    #select()
    print(h3.find_next("div").select("span.country-population")[0].get_text(strip = True))
    #select_one()
    print(h3.find_next("div").select_one("span.country-population").get_text(strip = True))

    print(h3.attrs) # to access attributes
```

```python
# to extract data to list and perform operations on it
all_countries = []
for h3 in all_h3:
    name = h3.get_text(strip = True)

    population = h3.find_next("div").select("span.country-population")[0].get_text(strip = True)

    all_countries.append([name, population])
```

```python
all_countries
```

```python
import pandas as pd
df = pd.DataFrame(all_countries, columns = ["Name", "Population"])
```

```python
df.to_csv("cleaned_data/data.csv", index = False)
```
