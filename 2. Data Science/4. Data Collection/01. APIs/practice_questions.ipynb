```python
# Q - 1:
# Suppose your manager asks you to scrape the website https://books.toscrape.com/.

# You need to:

# Download the webpage using requests.
# Check whether the request was successful.
# Use a User-Agent header.
# Handle exceptions using try-except.
# Parse the HTML using BeautifulSoup.
# Extract:
# Website title
# First book title
# Price of the first book
# Rating of the first book
# Image URL of the first book
# Print all book titles available on the first page.
# Save the HTML into a file called books.html.
import requests
from bs4 import BeautifulSoup
url = "https://books.toscrape.com/"

headers = {
    "User-Agen" : "Mozilla/5.0"
}

try:
    response = requests.get(url, headers = headers, timeout = 10)
    response.raise_for_status()

    # save html
    with open("scraped_data/books.html", "w", encoding="utf-8") as f:
        f.write(response.text)

    soup = BeautifulSoup(response.text, "html.parser")

    #Website title
    print("Website Title: ")
    print(soup.title.get_text())

    # First book
    first_book = soup.find("article")
    title = first_book.find("h3").find("a")["title"]
    price = first_book.find("p", class_="price_color").get_text()
    rating = first_book.find("p")["class"][1]
    image = first_book.find("img")["src"]

    print("\n First Book Details: ")
    print("Title : ", title)
    print("Price : ", price)
    print("Rating : ", rating)
    print("Image : ", image)

    print("\n All Books")

    books = soup.find_all("h3")

    for i, book in enumerate(books, start = 1):
        print(i, "-", book.find("a")["title"])

except requests.exceptions.RequestException as e:
    print("Error:", e)
```
