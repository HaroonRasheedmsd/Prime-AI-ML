```python
# Link of Books data
import requests
URL = "https://stephen-king-api.onrender.com/api/books"

# to send request we are using request method
req = requests.get(URL)
req
```

```python
json_data = req.json()
```

```python
json_data
```

```python
# To convert into PandaFrame
import pandas as pd
#What is json_normalize? : It is a built-in Pandas semi-structured data normalizer.The Function: It evaluates semi-structured JSON data and 
# flattens it into a 2D table (rows and columns
df = pd.json_normalize(json_data["data"]) 
df
```

```python
# To filter data 
df = df[['id', 'Year', 'Title', 'ISBN']]
df
```
