```python
# 2. bar(): a bar plot (or bar chart) is a graph that displays categorical data using rectangular bars. 
# each bar's height (or length, in horizental bars ) represents a value.
movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M
plt.bar(years, oscar_revenue, color = "grey")
plt.title("Revenue in each Year for Oscar movies")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

# plt.text() # it is used print label or values means print details and sepecifications
# like: text(x-axis position, y-axis position, text, alignment) 
for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i] + 10, str(oscar_revenue[i]), ha = "center")

# extend graph limits(area) using xlim(), and ylim()
plt.ylim(0, max(oscar_revenue) + 100)
#plt.xlim(min(years), max(years) + 1)
```

```python
# Scatter plots: we can get: data corelation, outliers can be detected, create clusters
# a scatter plot shows the relationship between two variables by plotting points on a 2D plane.
people = ["Person A", "Person B", "Person C", "Person D", "Person E", "Person F", 
          "Person G", "Person H", "Person I", "Person J"]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

# customization like color, size, marker, transparency, colormap
# for multiple different color
#colors = ["red", "yellow", "orange", "black", "purple","red", "yellow", "orange", "black", "purple"]

# in better way we use some logic
colors = ["green" if 110< x <130 else "red" for x in bp] 

#plt.scatter(age, bp, color = colors, s = bp, marker = ">", alpha = 0.8) # s for size, alpha for transparency value from 0 to 1. 

# for Color Map
plt.scatter(age, bp, s = bp, cmap = "plasma", c = bp) # cmap is used for colormap and there are multiple you can check on website
# "c" is the range of values that shows where perticular value be ploted

# To scale colors we use colorbar
plt.colorbar(label = "BP")

# plt.scatter(age, bp, color = "red") # for single color
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()

# Add Annotations: show exactly   point represent what values 
for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]), xytext = (age[i] + 1, bp[i] + 1)) # annotate(text, value , position)

plt.xlim(min(age), max(age) + 10)
plt.ylim(min(bp), max(bp) + 5)
```

```python
# Pie Charts: A pie chart is a circular chart divided into slices to show relative propertions of a whole
# each slice = a category, size of slice = value of that category relative to total
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

#plt.pie(amounts, labels = expenses)
#plt.style.available
#plt.style.use("classic")

# Customization
# 1. autopct ="%1.1f%%". shows percentage of each slice
# 2. colors i.e:
# 3. Wedgeprops: define (edgecolor, linewidth, linestyle, alpha, fill...) for a single slice
# 4. explode: use to highlight slice my be single or many
# 5. startangle : to rotate the pie chart
# 6. shadow: adds 3D effect or makes the chart circular
explode = [0,0,0,0.2,0]
colors = ["red", "orange", "pink", "blue", "yellow"]
plt.pie(amounts, labels = expenses, autopct = "%1.1f%%", colors = colors,
       wedgeprops = {
           "edgecolor": "black",
           "linewidth":1,
           "linestyle":"--",
           ## alpha (transparency) and fill for false or true like colors
       },
        explode = explode,
        startangle = 90,
        shadow = True
    )

#plt.style.use("default")
```
