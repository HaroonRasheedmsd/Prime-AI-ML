```python
import json

def loads_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
data = loads_data("customer_feedback_data.json")
print(data)
```

```python
# To Clean Data
def clean_data(data):
    # convert ratings from string to numbers
    text_to_num = { 'one':1, 'two':2, 'three':3, 'four':4, 'five':5, 'twenty six': 26}
    for user in data:
        raw_rating = user.get("rating")
        if isinstance(raw_rating, str):
            raw_rating = raw_rating.strip().lower()
            
            if raw_rating in text_to_num:
                raw_rating = text_to_num[raw_rating]
            elif raw_rating.isdigit():
                raw_rating = int(raw_rating)
            elif raw_rating == "":
                raw_rating = None
        user["rating"] = raw_rating

        # convert ages into integers

        # 
    return data    
```
