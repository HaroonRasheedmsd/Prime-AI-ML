```python
# Seaborn
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme() # before we starting with seaborn we select one theme here default one is selected
```

```python
sns.get_dataset_names() # Builtin data sets
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
print(tips.head())
```

```python
# lets create one relational plot
#sns.set_theme(style="whitegrid")
#sns.set_theme(style="darkgrid")
#sns.set_theme(style="white")
#sns.set_theme(style="dark")
sns.set_theme(style="ticks")
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "smoker",
    size = "size",
    style = "smoker"
)
```

```python
# scatter plot
tips = sns.load_dataset("tips")

sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "time"
)
```

```python
# line plot
flights = sns.load_dataset("flights")

sns.lineplot(
    data = flights,
    x = "year",
    y = "passengers"
)
```

```python
# Categorical plots: bar plots  and box plots
# print(tips.head())

# bar plot
sns.barplot(
    data = tips, 
    x = "day",
    y = "tip",
    hue = "sex"
)
```

```python
# box plot 
sns.boxplot(
    data = tips,
    x = "day",
    y = "tip"
)
```

```python
# Distribution Plots : histogram
penguins = sns.load_dataset("penguins")
sns.histplot(
    data = penguins,
    x = "body_mass_g",
    bins = 30
)
```

```python
# Matrix plots: Heatmaps- plot rectangular data as color encoded matrix 

# convert flight data into 2D array using pivot
flights_pivot = flights.pivot(index = "month", columns = "year", values = "passengers")
sns.heatmap(
    data = flights_pivot,
    cmap= "coolwarm",
    annot = True,
    fmt = "d"
)

plt.title("Passengers Heatmap")
plt.tight_layout()
```

```python
# Best practice of Seaborn : always use seaborn first if any customization them do that through Matplotlib
fig, ax = plt.subplots()
sns.lineplot(
    data = tips,
    x = "day",
    y = "total_bill",
    hue = "sex",
    marker = 'o',
    ax = ax,
    errorbar = None # to make more clear old one was ci = None
)

ax.set_title("Avg bill value for each day")
ax.set_xlabel("Days")
ax.set_ylabel("Total Bill Amount")
fig.tight_layout()
```
