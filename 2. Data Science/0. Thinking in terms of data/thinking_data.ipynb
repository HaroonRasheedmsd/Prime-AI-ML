Assignment-1  To Clean and structure data

```python
# clrean & structure the Data
def clean_data(data):
    text_to_num = {"one":1, "two":2, "three":3, "four":4, "five":5, "twenty four": 24}

    cleaned_data = []
    seen = set()

    for user in data: 

        #  ===== clean rating =====
        rating = user.get("rating")
        if isinstance(rating, str):
            rating = rating.strip().lower()

            if rating in text_to_num:
                rating = text_to_num [rating]
            elif rating.isdigit():
                rating = int(rating)
            elif rating == "":
                rating = None

        user["rating"] = rating

        # ========== clean age ==========
        age = user.get("age")
        if isinstance(age, str):
            age = age.strip().lower()
            
            if age in text_to_num:
                age = text_to_num[age]
        
            elif age.isdigit():
                age = int(age)
            
        user["age"] = age

        # ======== Handle missing rating ======
        if user.get("rating") is None:
            user["rating"] = 0

        # ===== Remove duplicates ==========
        key = (user["name"], user["age"])

        if key not in seen:
            seen.add(key)
            cleaned_data.append(user)

        # remove duplication another logic
        if(user["name"].strip() in seen):
            continue
        cleaned_data.append(user)
    return cleaned_data
```

```python
data = clean_data(data)
print(data)
```

```python
# Average ratinga and Poor rating bellow 3
def get_insights(data):
     # avg rating
    ratings = [
        user["rating"]
        for user in data
        if user["rating"] is not None
    ]

    # this is equavalent to the above list of rating
    # ratings = []
    # for user in data:
    #     if user["rating"] is not None:
    #         ratings.append(user["rating"])

    avg_rating = sum(ratings) / len(ratings)

    print("Average Rating: ", round(avg_rating, 2))

        # poor rating
    poor_ratings = 0
    for user in data:
        if user["rating"] is not None and user["rating"] < 3:
            poor_ratings += 1

    percentage = (poor_ratings / len(data)) * 100
    print(f"Poor rating Percentage: {percentage:.2f}%")
```
