```python
#Start by loading the IRIS dataset
import pandas as pd
data = pd.read_csv("IRIS.csv")


# Q - 1

# A. Display the following information:
# First 10 rows
data.head(10)
# Shape & Data types
print(data.shape, data.dtypes)
# Summary statistics (mean, std, min, max)
#data.describe()
#or
print(data[["sepal_length", "sepal_width", "petal_length", "petal_width"]].agg(["mean", "std", "min", "max"]))

# B. Select only those rows where: • petal_length > 4.5,  • species = "Iris-virginica"
# petal_length > 4.5
#data.query(' petal_length > 4.5 and species == "Iris-virginica"') 
#or
data[(data["petal_length"] > 4.5) & (data["species"] == "Iris-virginica")]

#C. Group by species and compute: • Average sepal_length, • Maximum petal_width, • Standard deviation of sepal_width
result = data.groupby("species").agg(
    avg_sepal_length = ("sepal_length", "mean"),
    max_petal_width = ("petal_width", "max"),
    std_sepal_width = ("sepal_width", "std")
)
print(result)

#D. Create a new column: "petal_ratio" = petal_length / petal_width, • For each species, find the average petal_ratio.
# Create new column
data["petal_ratio"] = data["petal_length"] / data["petal_width"]

# Average petal_ratio for each species
avg_ratio = data.groupby("species")["petal_ratio"].mean()

print(avg_ratio)
```

```python
# Q - 2 Q2. Start by loading the TITANIC dataset. 
import pandas as pd

titanic = pd.read_csv("Titanic-Dataset.csv")

# A. Display only columns: Name, Sex, Age, Fare, Survived
titanic[["Name", "Sex", "Age", "Fare", "Survived"]]

# B. Select passengers who:, • Are female, • Fare > 30
female_passengers = titanic[
    (titanic["Sex"] == "female") &
    (titanic["Fare"] > 30)
]

print(female_passengers)

# or 
# titanic.query('Sex == "female" and Fare > 30')

#C. Group by Pclass and compute:, • Survival rate, • Average Fare, • Average Age
result = titanic.groupby("Pclass").agg(
    survival_rate=("Survived", "mean"),
    average_fare=("Fare", "mean"),
    average_age=("Age", "mean")
)

print(result)
```
