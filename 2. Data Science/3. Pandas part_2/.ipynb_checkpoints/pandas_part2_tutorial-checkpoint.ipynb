```python
# Transformin Data (Feeture Engineering) : to create new columns, to create some new features that can improve the performance of the models

# e.g: we need "tax" column based on salary like:
import pandas as pd
df = pd.read_csv("employee_data.csv")
df1 = df.copy()
df1["tax"] = df1["Salary"].apply(lambda x : "20%" if x >=60000 else "10%") # apply() function is used and it is higher order function as we passed another function as a parameter
df1

# map() : to map existing values with new values: like Male => M, Female=> F, etc using dictionary
department_map = {"Finance" : "F", "Marketing" : "M", "HR" : "HR", "IT" : "IT"}
df1["Department"] = df1["Department"].map(department_map)
df1

# assing(): we create new column by assigning them some values like: Salary increase by 10% 
df1.assign(new_income = df1["Salary"] * 1.1)
df1
```
