```python
# 3. Pandas Started. we use Pandas to play with data. we use NumPy for numerical computation (provide ndarray DS) but 
# Pandas are are mainly used for Data manipulation and Provide Data Structure like. Dataframe, and Series 
# 1. Series: 1D labelled array     2. Dataframe: 2D labelled array
import pandas as pd

# Convert Python Dictionary to Pandas DataFrame
info = {
    "name" : ["Ali", "Haroon", "Khan"],
    "CGPA" : [3.33, 3.22, 4.0]
}

df = pd.DataFrame(info)
print(df)
print("\n\nProper Table form is")
df
```

```python
# 4. Series: 1D labelled array. it is also homogenous means same type of data are stored  
s = pd.Series([23,24,25,26], index = ["Ali", "Harry", "Khan", "Jhon"]) # here index = [] used to labelled series data
# without index = [] by default 0,1,2,3,... labels are asign
print(s)
print(s["Ali"]) # acces value using label
print(s["Jhon"]) 
print(s[0]) # acces value using index
print(s.index) # Print Labels of series
print(type(s))
```

```python
# 5. Properties of Serires: i.Homogenous, ii.Vectorized operation, iii.Handle missing values with NaN, 
    # iv.Mutable values but immutable size
# i.Homogenous example
s = pd.Series([23,24,25,26, "abc"], index = ["Ali", "Harry", "Khan", "Jhon", 23])
print(s)
print(type(s))

# ii. Vectorized operation eg.
s1 = pd.Series([1,3,4,5,6])
s2 = pd.Series([11,33,44,55,66])
s3 = s1*3
print(s3)
print(s1 + s2)

# iv. Mutable Values
s1[0] = 40
print(s1)

# iv. Immutable size if we change, that create new series changes happen there original series remain same e.g.
changed_s1 = s1.drop(0) # this drop remove index
print(s1)
print(changed_s1)
```

```python
# AQI Data Set
df = pd.read_csv("globalAirQuality.csv")
#df.head(10)
#df.tail()
df.describe()
```

```python
# Selecting Data
# Column wise data access
df["country"]
df[["city", "aqi"]]

# for rows
# Select rows: for rows we use i.loc for labels and ii.iloc for indexes 
df.loc[0] # first row data 
df.loc[0:3] # here we use slice format from row 0 to 3: loc[starting : ending ] also ending index included (inlusive)

df.iloc[2] # here index 2 will be printed
df.iloc[0:2] # In indexing the ending value in exclusive not included
# print row and column using loc and iloc
df.loc[0, "aqi"]
df.loc[0,["city", "aqi"]]
df.loc[0:2,["city", "aqi"]]

#df.iloc[0:2, "aqi"] # this will cause and error as iloc don't accept string as it work with indexes so only numerics are allow for this first we will print index
df.columns # this will print columns and there we will count name of column and will be selected
df.iloc[0:2, 2]
df.iloc[0:2, 1:5]

# cell: we use at for labels and iat for index
df.at[0, "city"] # 0 shows row and city is the label
df.iat[0, 2] # here 0 shows row and 2 shows index of city

# * one thing important that we get view only not copy of that.
# so we do mistake that we replace the data by mistake. so if we need to change something firs make copy using copy() function at last connect with dot(.)
# so that we make safe room. then we change
# e.g.
cities = df["city"].copy()
cities[0] = "Mumbai" # here copy is updated
cities
```

```python
# Filtering of Data 
# 1.using [] indxeing or labels
df[df["aqi"] > 100]

df[(df["aqi"] > 100) & (df["temperature"] > 30)]

# here done chaining along with filtering
df[df["aqi"] > 100 ] [["city",  "aqi"]]

# Query method return copy not view
# 2.Using query() . Rules: we write condintion in string, using column name , backticks for col_name with space or sepcial chars, operators are same and also refrence to python variable, and Chained comparision
df.query("aqi > 100")
df.query("aqi > 100 & temperature >30")

# also we can do chaining where only needed column can be filtered
df.query("aqi > 100 & temperature >30")[["city", "aqi"]]
df.query("aqi > 100 & temperature >30")[["city", "aqi"]]

aqi_val = 100
df.query("aqi > @aqi_val & temperature >30")[["city", "aqi"]] # python refrence variable
```

```python
# CLeaning Data
df = pd.read_csv("raw_data.csv")
df
# 1. Handle missing Values
df.isnull() # it return True if any null or missing value found
df.isna() # it is the alias(second name or same function) of isnull() 
df.isnull().sum() # this is chaining like it tell how many values are missing in any column 
df.dropna() # this will drop any row where any missing value occur
df.dropna(axis = 1) # this will drop column if any missing value occur. so only axis changed into column
df.fillna(0) # this will fill all missing values with value we passed here we passed 0

# to fill specific column with specific value and to seperate clean data 

cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean) # this will fill only specific column we selected
cleaned_data

# ffill() : forward fill here wi move vertically and if any missing value occure in the upcomming cell we fill with 
# that value like in row 0 value 1 but row 1 is null we fill with row 0 value called 1
df.ffill()

# bfill() : it is just opposite of the ffill()
df.bfill()
```

```python
# Handle Duplicate values, so we deduplication to remove duplicate values
df.duplicated() # this will print True if any duplicated record found 
df["country"].duplicated() # only for one column
df[["country" , "gender"]].duplicated() # only for one column

df2 = df.copy()
df2.drop_duplicates(inplace = True) # inplace is used that values change oringinally not only change in view that is  why i made the copy also 
```

```python
# Handle Different Data Types & Date-time
# to check data type used dtypes function 
df.dtypes 
df2.dtypes
df2 = df2.fillna(0)
df2["age"] = df2["age"].astype("int64")
df2.dtypes
```
