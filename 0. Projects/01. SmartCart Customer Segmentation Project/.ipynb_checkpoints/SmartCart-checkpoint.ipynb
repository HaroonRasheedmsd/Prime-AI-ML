# Data Preprocessing

## 1.Handle Missing values

```python
df["Income"] = df["Income"].fillna(df["Income"].median())
```

```python
df.isnull().sum()
```

## Feature Engineering

```python
# here we seen two dates column that needs feeturing engineering

## 1.Age : needs to add Age column to replace year of birth which is not effective
df["Age"] = 2026-df["Year_Birth"]
```

```python
df.head()
```

```python
## 2.Customer_Tenure_Days: need to add Customer_Tenure_Days column as clutering algo are not good to 
## to read or use these type of format like m-d-y|d-m-y, for we should subtract all date with reference date

df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], dayfirst=True)
reference_date = df["Dt_Customer"].max()
df["Customer_Tenure_Days"] = (reference_date - df["Dt_Customer"]).dt.days
```

```python
df.head()
df.columns
```

```python
## 3.Spending: add Spending column that contain total spend

df["Total_Spending"] = df["MntWines"] + df["MntFruits"] + df["MntMeatProducts"] + df["MntFishProducts"] + df["MntSweetProducts"] + df["MntGoldProds"]
```

```python
## 4.Total_Children: add Total_Children column

df["Total_Children"] = df["Kidhome"] + df["Teenhome"]
```

```python
## Now we have some columns containing multiple categories lets work on that

### i.Education

df["Education"].value_counts()

# so lets make 3 Categories: 
## Undergraduate(2n Cycle, Basic), Graduate(Graduation), Postgraduate(PhD, Master)

df["Education"] = df["Education"].replace({
    "Basic": "Undergraduate",
    "2n Cycle": "Undergraduate",
    "Graduation": "Graduate",
    "Master": "Postgraduate",
    "PhD": "Postgraduate"
})
```

```python
### ii.Marital_Status

df["Marital_Status"].value_counts()

df["Living_With"] = df["Marital_Status"].replace({
    "Married": "Partner",
    "Together": "Partner",
    "Single": "Alone",
    "Divorced": "Alone",
    "YOLO": "Alone",
    "Absurd": "Alone",
    "Widow": "Alone"
})

df["Living_With"].value_counts()
```

# Drop Columns

```python
df.head()
```

```python
cols = ["ID", "Year_Birth", "Marital_Status","Kidhome", "Teenhome", "Dt_Customer"] 
spending_cols = ["MntWines", "MntFruits", "MntMeatProducts", "MntFishProducts", "MntSweetProducts", "MntGoldProds"] 

cols_to_drop = cols + spending_cols

df_cleaned = df.drop(columns = cols_to_drop)
```

```python
df_cleaned.shape
df.columns
```

# Outliers

```python
# to detect outliers we generate mulitple plots of our data

cols = ["Income", "Recency", "Response", "Age", "Total_Spending", "Total_Children"]

# relative plots of some features- pair plots
sns.pairplot(df_cleaned[cols])
```

```python
# # Remove Outliers

print("Data size with Outliers: ", len(df_cleaned))

df_cleaned = df_cleaned[ (df_cleaned["Age"] < 90 )]
df_cleaned = df_cleaned[ (df_cleaned["Income"] < 600_000)]

print("Data size without Outliers: ", len(df_cleaned))
```

# Encoding

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()
cat_cols = ["Education", "Living_With"]

enc_cols = ohe.fit_transform(df_cleaned[cat_cols])
```

```python
enc_df = pd.DataFrame(enc_cols.toarray(), columns=ohe.get_feature_names_out(cat_cols), index=df_cleaned.index)
```

```python
df_encoded = pd.concat([df_cleaned.drop(columns = cat_cols), enc_df], axis=1)
```

# Analyze K Value
## 1.Elbow Method

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator

wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
```

```python
knee = KneeLocator(range(1, 11), wcss, curve="convex", direction="decreasing")
optimal_k = knee.elbow
```

```python
print("Best k =", optimal_k)
```

```python
# plot elbow
plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel("K")
plt.ylabel("WCSS")
```

## 2. Silhouete Score

```python
from sklearn.metrics import silhouette_score

scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_pca)
    score = silhouette_score(X_pca, labels)
    scores.append(score)

# plot silhouette
plt.plot(range(2, 11), scores, marker='o')
plt.xlabel("K")
plt.ylabel("Silhouette Score")
```
